Splitting the documents into chunks

```python
# Split into manageable chunks
text_splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=200)
splits = text_splitter.split_documents(pdf_docs)
print(f"Generated {len(splits)} chunks.")
```

Vector RAG setuo using FAISS vector database.

```python
# ==========================================
# 3. Setup Vector RAG
# ==========================================
print("Building Vector Store (FAISS)...")
vectorstore = FAISS.from_documents(splits, embedder)
retriever = vectorstore.as_retriever()
```

Instructing the LLM to be annergy

```python
# Create a standard document retrieval chain
system_prompt = (
    "You are an expert energy analyst. Use the given context to answer the question. "
    "If you don't know the answer, say so.\n\nContext:\n{context}"
)
prompt = ChatPromptTemplate.from_messages([
    ("system", system_prompt),
    ("human", "{input}"),
])
question_answer_chain = create_stuff_documents_chain(llm, prompt)
vector_rag_chain = create_retrieval_chain(retriever, question_answer_chain)
```

```python
# ==========================================
# 4. Setup Graph RAG (Neo4j Aura)
# ==========================================
graph = Neo4jGraph(
url="neo4j+s://=.databases.neo4j.io",
username="=", password="",database="")


# Extract nodes and relationships using LLM
llm_transformer = LLMGraphTransformer(
    llm=llm,

allowed_nodes=["BasinRegion","DisruptionEvent","ReportOutlook","MarketMetricForecast","CommodityEnergy","InfrastructureChokepoint",
                   "GeopoliticalEntity","OilField","PortTerminal","Vessel","PetroleumProduct","Policy","FinancialContract","ChemicalProperty",
                   "Organization", "Location", "Commodity", "Process", "Metric","CrudeGrade", "Region", "Company", "Refinery", "MarketIndicator", "GeopoliticalEvent","Pipeline"],
allowed_relationships=["ALTERS_GENERATION_MIX","EXTRACTS_TAX","DRIVES_INVENTORY_DRAW","PROJECTS_VALUE",
                           "GENERATES_BYPRODUCT","INCENTIVIZES_DRILLING","AGGREGATES_INTO","TRIGGERS_SHUTIN,DISRUPTS_FLOW","PRODUCES",
                           "AFFECTS","AFFECTS_SUPPLY_OF", "LOCATED_IN", "MEASURED_IN","PRODUCED_IN", "IMPACTS_PRICE","SUPPLIES_TO",
                           "FORECASTS_OUTLOOK","EXTRACTED_FROM","OPERATES","PRICED_AGAINST","ENACTS","SUBJECT_TO","DISRUPTS","HAS_PROPERTY","OPERATED_BY","REFINED_INTO","TRANSPORTED_VIA"]
```

Extract graph documents and add to Neo4j

```python
# Extract graph documents and add to Neo4j
graph_documents = llm_transformer.convert_to_graph_documents(splits)
```

```python
graph.add_graph_documents(graph_documents)
```

Printing the schema

```python
graph.refresh_schema()
print(graph.schema)
```

Creating the question and answer chain

```python
# Create the Graph QA Chain
# To this:
graph_rag_chain = GraphCypherQAChain.from_llm(
    llm=llm,
    graph=graph,
    verbose=False,
    allow_dangerous_requests=True
)
```

Function to run the benchmark of vector RAG and graph RAG

```python
# ==========================================
# 5. Benchmarking Function
# ==========================================
def benchmark_queries(queries):
    print("\n--- Starting Benchmark ---")

    for query in queries:
        print(f"\nQuery: '{query}'")

        # Benchmark Vector RAG
        start_time = time.time()
        vector_response = vector_rag_chain.invoke({"input": query})
        vector_time = time.time() - start_time

        print(f"[Vector RAG] Time: {vector_time:.2f}s")
        print(f"[Vector RAG] Answer: {vector_response['answer'][:200]}...") # Truncated


        # Benchmark Graph RAG
        start_time = time.time()
        graph_response = graph_rag_chain.invoke(query)
        graph_time = time.time() - start_time

        print(f"[Graph RAG] Time: {graph_time:.2f}s")
        print(f"[Graph RAG] Answer: {graph_response['result'][:200]}...") # Truncated

    return vector_response, graph_response
```
