```python
from neo4j import GraphDatabase

# Replace with your Neo4j credentials
URI = "neo4j+s://=.databases.neo4j.io"
AUTH = ("", "")

def fetch_graph_data(driver):
    # We limit the query to prevent crashing memory on massive KGs.
    # Adjust or remove LIMIT based on your graph size.
    query = """
        MATCH (n)-[r]->(m)
        RETURN id(n) AS source, labels(n)[0] as source_label,
              id(m) AS target, labels(m)[0] as target_label
        LIMIT 5000
        """
    with driver.session() as session:
        result = session.run(query)
        return [record.data() for record in result]

driver = GraphDatabase.driver(URI, auth=AUTH)
edge_data = fetch_graph_data(driver)
driver.close()
```

Creating sources and targets.

```python
import networkx as nx

G = nx.Graph() # Undirected graph

for record in edge_data:
    src = record["source"]
    tgt = record["target"]

    # Add nodes with their Neo4j labels as attributes
    G.add_node(src, label=record["source_label"])
    G.add_node(tgt, label=record["target_label"])

    # Add the edge
    G.add_edge(src, tgt)

print(f"Graph loaded: {G.number_of_nodes()} nodes, {G.number_of_edges()} edges")
```

```python
from neo4j import GraphDatabase
import networkx as nx
import pandas as pd
import numpy as np

# Replace with your Neo4j credentials
URI = "neo4j+s://.databases.neo4j.io"
AUTH = ("", "")

cypher_query = """
MATCH (s)-[r]->(t)
RETURN id(s) AS src_id,
       coalesce(s.name, labels(s)[0] + "_" + toString(id(s))) AS src_name,
       labels(s)[0] AS src_label,
       id(t) AS tgt_id,
       coalesce(t.name, labels(t)[0] + "_" + toString(id(t))) AS tgt_name,
       labels(t)[0] AS tgt_label,
       type(r) AS rel_type
"""

driver = GraphDatabase.driver(URI, auth=AUTH)
with driver.session() as session:
    records = session.run(cypher_query).data()
driver.close()

# 2. Construct Directed Graph
G = nx.DiGraph()
for rec in records:
    G.add_node(rec["src_name"], label=rec["src_label"])
    G.add_node(rec["tgt_name"], label=rec["tgt_label"])
    G.add_edge(rec["src_name"], rec["tgt_name"], relation=rec["rel_type"])

# 3. Compute Directed Degree Metrics
in_degrees = dict(G.in_degree())
out_degrees = dict(G.out_degree())
total_degrees = dict(G.degree())

# 4. Assemble Results DataFrame
df_degree = pd.DataFrame([
    {
        "entity": node,
        "category": G.nodes[node].get("label", "Unknown"),
        "total_degree": total_degrees[node],
        "in_degree": in_degrees[node],
        "out_degree": out_degrees[node],
        # Positive asymmetry = Sink (In-dominated), Negative = Source (Out-dominated)
        "asymmetry_ratio": (in_degrees[node] - out_degrees[node]) / (total_degrees[node] if total_degrees[node] > 0 else 1)
    }
    for node in G.nodes()
]).sort_values(by="total_degree", ascending=False)

print(df_degree.head(10).to_string(index=False))
```

```python
import matplotlib.pyplot as plt
import networkx as nx
from neo4j import GraphDatabase


class AuraGraphExtractor:

  def __init__(self, uri, user, password):
    self.driver = GraphDatabase.driver(uri, auth=(user, password))

  def close(self):
    self.driver.close()

  def get_subgraph(self, cypher_query, parameters=None):
    """Executes a Cypher query and constructs a NetworkX Directed Graph."""
    G = nx.DiGraph()

    with self.driver.session() as session:
      result = session.run(cypher_query, parameters or {})
      for record in result:
        source_node = record["source"]
        target_node = record["target"]
        rel = record["rel"]

        # Unique node identifiers
        src_id = source_node.element_id
        tgt_id = target_node.element_id

        # Node labels and properties
        src_label = list(source_node.labels)[0] if source_node.labels else "Node"
        tgt_label = list(target_node.labels)[0] if target_node.labels else "Node"

        # Add source and target nodes with attributes
        G.add_node(
            src_id,
            display_name=source_node.get("name", source_node.get("id", src_id)),
            label=src_label,
            **dict(source_node),
        )
        G.add_node(
            tgt_id,
            display_name=target_node.get("name", target_node.get("id", tgt_id)),
            label=tgt_label,
            **dict(target_node),
        )

        # Add directed edge with relationship type and attributes
        G.add_edge(src_id, tgt_id, type=rel.type, **dict(rel))

    return G
```

```python
import matplotlib.pyplot as plt
import networkx as nx
from neo4j import GraphDatabase


class AuraFullGraphExtractor:

  def __init__(self, uri, user, password):
    self.driver = GraphDatabase.driver(uri, auth=(user, password))

  def close(self):
    self.driver.close()

  def get_entire_graph(self):
    """Fetches all nodes and relationships in the database into a NetworkX DiGraph."""
    G = nx.DiGraph()

    # Query matches isolated nodes as well as connected pairs
    cypher_query = """
        MATCH (source)
        OPTIONAL MATCH (source)-[rel]->(target)
        RETURN source, rel, target
        """

    with self.driver.session() as session:
      result = session.run(cypher_query)
      for record in result:
        source = record["source"]
        rel = record["rel"]
        target = record["target"]

        # Process Source Node
        src_id = source.element_id
        src_label = (
            list(source.labels)[0]
            if getattr(source, "labels", None)
            else "Node"
        )

        G.add_node(
            src_id,
            display_name=source.get(
                "name", source.get("title", source.get("id", src_id))
            ),
            label=src_label,
            **dict(source),
        )

        # Process Target Node and Relationship (if present)
        if target is not None and rel is not None:
          tgt_id = target.element_id
          tgt_label = (
              list(target.labels)[0]
              if getattr(target, "labels", None)
              else "Node"
          )

          G.add_node(
              tgt_id,
              display_name=target.get(
                  "name", target.get("title", target.get("id", tgt_id))
              ),
              label=tgt_label,
              **dict(target),
          )

          G.add_edge(src_id, tgt_id, type=rel.type, **dict(rel))

    return G
```
